```python
import pandas as pd
import re
from gensim.models import Word2Vec
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# 1. Load dataset
df = pd.read_csv("Datasets/CBOW/twitter_dataset.csv")

# 2. Preprocess tweets
stop_words = set(ENGLISH_STOP_WORDS)

def preprocess(text):
    words = re.findall(r'\b[a-z]+\b', str(text).lower())
    return [w for w in words if w not in stop_words]

sentences = df["Text"].fillna("").apply(preprocess).tolist()

# 3. Train CBOW Word2Vec model
model = Word2Vec(sentences, vector_size=100, window=5,
                 min_count=1, sg=0, workers=4, epochs=10)

# 4. Evaluate the model
print("Vocabulary Size:", len(model.wv))
print("Word Embedding:", model.wv["party"])
print("Similar Words:", model.wv.most_similar("party"))

# 5. Find similarity between two words
print("Similarity:", model.wv.similarity("party", "style"))
```

```python
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

df = pd.read_csv("Datasets/CBOW/twitter_dataset.csv")

stop_words = set(ENGLISH_STOP_WORDS)

def preprocess(text):
    words = re.findall(r'\b[a-z]+\b', text.lower())
    return [w for w in words if w not in stop_words]

sentences = df["Text"].fillna("").apply(preprocess).tolist()

w2v = Word2Vec(sentences, vector_size=100, window=5, epochs=10, workers=4, min_count=2, sg=0)

print("Vocabulary Words: ", len(w2v.wv))
print("Word Embedding: ", w2v.wv["party"])
print("Similar Words: ", w2v.wv.most_similar("party"))

print("Similarity: ", w2v.wv.similarity("party", "style"))
```
