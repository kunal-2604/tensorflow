```python
import pandas as pd
import re
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# 1. Load and preprocess tweets
df = pd.read_csv("Datasets/CBOW/twitter_dataset.csv")

def preprocess(text):
    return re.findall(r'\b[a-z]+\b', str(text).lower())

sentences = df["Text"].apply(preprocess).tolist()

# 2. Train CBOW Word2Vec model
model = Word2Vec(sentences, vector_size=100, window=5,
                 min_count=1, sg=0, workers=4, epochs=100)

# 3. Convert each tweet into an average word embedding
def get_vector(words):
    vectors = [model.wv[word] for word in words if word in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(100)

tweet_vectors = np.array([get_vector(words) for words in sentences])

# 4. Get user input and calculate similarity
query = input("Enter your search text: ")
query_vector = get_vector(preprocess(query)).reshape(1, -1)

similarity = cosine_similarity(query_vector, tweet_vectors)[0]

# 5. Recommend the most similar tweets
df["Similarity"] = similarity
print(df.sort_values("Similarity", ascending=False)[["Tweet_ID", "Text", "Similarity"]].head(20))
```

```python
import pandas as pd
import numpy as np
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import re

df = pd.read_csv("Datasets/CBOW/twitter_dataset.csv")

def preprocess(text):
    return re.findall(r'\b[a-z]+\b', text.lower())

sentences = df["Text"].fillna("").apply(preprocess).tolist()

w2v = Word2Vec(sentences, vector_size=100, min_count=1, window=5, sg=0, workers=4, epochs=100)

def doucment_vector(words):
    vectors = [w2v.wv[w] for w in words if w in w2v.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(100)

tweets = np.array([doucment_vector(s) for s in sentences])

query = input("Enter your query: ")
query_vector = doucment_vector(preprocess(query)).reshape(1, -1)

cosine_similarity = cosine_similarity(query_vector, tweets)[0]

df["Similarity"] = cosine_similarity
print(df.sort_values("Similarity", ascending=False)[["Tweet_ID", "Text", "Similarity"]].head(20))
```
