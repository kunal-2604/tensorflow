```python
import re
from gensim.models import Word2Vec

# 1. Read document
with open("Datasets/CBOW/document3.txt", "r") as file:
    text = file.read()

# 2. Data preparation
words = re.findall(r'\b[a-z]+\b', text.lower())
sentences = [words]

# 3. Train CBOW Word2Vec model
model = Word2Vec(sentences, vector_size=100, window=5,
                 min_count=1, sg=0, workers=4, epochs=100)

# 4. Evaluate word embeddings
print("Vocabulary:", list(model.wv.index_to_key))

word = "important"
print("\nEmbedding of", word, ":\n", model.wv[word])

print("\nSimilar words:")
print(model.wv.most_similar(word))
```

```python

# Step 1: Import libraries
import re
import numpy as np
import matplotlib.pyplot as plt

from gensim.models import Word2Vec

# Step 2: Load the given Document 3
with open("Datasets/CBOW/document3.txt", "r", encoding="utf-8") as file:
    document = file.read()

# Step 3: Preprocess the document
# Convert text into lowercase
document = document.lower()

# Split document into sentences
sentences = re.split(r'(?<=[.!?])\s+', document)

# Tokenize sentences into words
def preprocess_text(sentence):
    words = re.findall(r"[a-z]+(?:-[a-z]+)*", sentence)
    return words

# Convert sentences into lists of words
tokenized_sentences = [
    preprocess_text(sentence) for sentence in sentences
]

# Remove empty sentences
tokenized_sentences = [
    words for words in tokenized_sentences if len(words) > 0
]

print("Tokenized Document:")
print(tokenized_sentences)

# Step 4: Train Word2Vec CBOW model
# sg=0 selects CBOW architecture
cbow_model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    workers=1,
    epochs=100
)

# Step 5: Display vocabulary
print("\nVocabulary Size:", len(cbow_model.wv))

print("\nVocabulary Words:")
print(list(cbow_model.wv.index_to_key))

# Step 6: Display word embeddings
word = "influenza"

print("\nSelected Word:", word)
print("Word Embedding:")
print(cbow_model.wv[word])

# Step 7: Find similar words
print("\nMost Similar Words:")
similar_words = cbow_model.wv.most_similar(word, topn=5)
```

```python
import re
from gensim.models import Word2Vec

with open("Datasets/CBOW/document3.txt", 'r', encoding='utf-8') as file:
    doc = file.read()

doc = doc.lower()

sentences = re.split(r'(?<=[.!?])\s+', doc)

def preprocess(text):
    return re.findall(r'\b[a-z]+\b', text)

tokenized_sentences = [preprocess(s) for s in sentences]
tokenized_sentences = [s for s in tokenized_sentences if len(s) > 0]

print(tokenized_sentences)

w2v = Word2Vec(tokenized_sentences, vector_size=100, workers=4, epochs=100, min_count=1, window=5)

print("Vocabulary Length: ", len(w2v.wv))
print("Vocabulary list: ")
print(list(w2v.wv.index_to_key))

word = "influenza"
print("Word Embedding: ", w2v.wv[word])

print("Similar Words: ", w2v.wv.most_similar(word, topn=5))
```
